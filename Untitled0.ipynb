```python
import pandas as pd
df = pd.DataFrame({
    "salary" : [40,50,60,70,80,90,200] # here 200 is outlier

})
```

```python
df.describe()
```

```python
Q1 = df["salary"].quantile(0.25)
Q3 = df["salary"].quantile(0.75)
print(Q1, Q3)
```

```python
iqr = Q3 - Q1
print(iqr)
```

```python
#IQR (Interquartile Range) tells the range of the middle 50% of data.
```

```python
# Q1 = lower middle value
# Q3 = upper middle value

#  IQR = difference between them
```

```python
outliers = df[(df["salary"]> Q3 + 1.5*iqr) | (df["salary"]< Q1 - 1.5*iqr) ] # formulae

```
